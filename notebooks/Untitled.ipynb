```python
df.rename(columns={
    "Number": "Policy_Number",
    "Name": "Name",
    "Age": "Age",
    "Premium Amount": "Premium_Amount",
    "Policy Type": "Policy_Type",
    "Mobile number": "Mobile_Number",
    "Gender": "Gender",
    "DOB": "DOB",
    "Premium Type": "Premium_Type",
    "Date_of_Commencemnet": "Date_of_Commencement",
    "End_of_Premium_Paying": "End_of_Premium_Paying",
    "Date_of_Maturity": "Date_of_Maturity"
}, inplace=True)
```

```python
print(df.columns.tolist())
```

```python
df["DOB"] = pd.to_datetime(df["DOB"], dayfirst=True)

df["Date_of_Commencement"] = pd.to_datetime(
    df["Date_of_Commencement"], dayfirst=True
)

df["End_of_Premium_Paying"] = pd.to_datetime(
    df["End_of_Premium_Paying"], dayfirst=True
)

df["Date_of_Maturity"] = pd.to_datetime(
    df["Date_of_Maturity"], dayfirst=True
)
```

```python
clean_df = df.copy()
```

```python
date_columns = [
    "DOB",
    "Date_of_Commencement",
    "End_of_Premium_Paying",
    "Date_of_Maturity"
]

for col in date_columns:
    clean_df[col] = pd.to_datetime(
        clean_df[col],
        dayfirst=True,
        errors="coerce"
    )
```

```python
for col in date_columns:
    print(f"\nInvalid dates in {col}:")
    print(clean_df[clean_df[col].isna()][["Policy_Number", col]])
```
